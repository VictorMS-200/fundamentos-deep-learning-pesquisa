## 4. Regras rápidas para nomes de variáveis

Um nome de variável:

- não deve conter espaços;
- não deve começar com número;
- pode usar letras, números depois do primeiro caractere e `_`;
- diferencia maiúsculas de minúsculas.

## 5. Exemplo guiado: verificando tipos com `type()`

```python
quantidade = 10
preco = 12.5
curso = "Python"
ativo = True

print(type(quantidade))
print(type(preco))
print(type(curso))
print(type(ativo))
```

Uma saída como `<class 'int'>` indica que o valor pertence ao tipo `int`. Nesta lição, concentre-se no nome do tipo: `int`, `float`, `str` ou `bool`.

## 8. `print()` com vírgula e `str()`

```python
idade = 20

print("Idade:", idade)

mensagem = "Idade: " + str(idade)
print(mensagem)
```

`print("Idade:", idade)` funciona porque `print()` pode receber vários valores separados por vírgula.

Já em `"Idade: " + str(idade)`, `str()` converte o número para texto antes da junção com `+`.

### Exercício

## 9. Comparando número e texto

```python
numero = 10
texto = "10"

print(numero, type(numero))
print(texto, type(texto))

print(10 + 5)
print("10" + "5")
```

### Observe

Os valores podem parecer semelhantes na tela, mas seus tipos mudam o comportamento das operações.

Escreva uma frase curta, na célula abaixo, descrevendo a diferença que você observou.

**Sua observação:** números dentro de aspas "", são tratados como str, ou como textos, ou seja, na última linha, ele tratou "10" + "5", como texto, o que deu como resultado 105, e não 15, como na penúltima, pois eles estavam entre aspas.
