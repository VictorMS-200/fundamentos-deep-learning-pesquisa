````python
dados = ["Ana", 20, 1.65, True]
print(dados)
````

Por exemplo:


Não é uma obrigação da linguagem. É uma escolha de organização.

## 13. Lista de dicionários

Podemos reunir pequenos registros em uma lista.


````python
amostras = [
    {"id": "A001", "classe": "classe_a"},
    {"id": "A002", "classe": "classe_b"}
]

print(amostras[0])
print(amostras[0]["id"])
print(amostras[1]["classe"])
````
