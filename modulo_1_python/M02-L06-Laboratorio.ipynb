## 22. Função para listar arquivos

Vamos organizar uma operação repetida em uma função.

# 23. Organização de um dataset simples

## 26. Erro adicional — separador incorreto

Considere um arquivo que usa `;`.

````python
csv_ponto_virgula = Path("dados_ponto_virgula.csv")

with open(csv_ponto_virgula, "w", encoding="utf-8") as f:
    f.write("nome;quantidade\n")
    f.write("classe_a;2\n")
````

````python
with open(csv_ponto_virgula, "r", encoding="utf-8", newline="") as f:
    leitor = csv.reader(f)
    for linha in leitor:
        print(linha)
````

Observe que, usando o separador padrão do leitor, a linha pode aparecer como **um único campo**.

Para este arquivo, o leitor pode receber `delimiter=";"`:

````python
with open(csv_ponto_virgula, "r", encoding="utf-8", newline="") as f:
    leitor = csv.reader(f, delimiter=";")
    for linha in leitor:
        print(linha)
````
