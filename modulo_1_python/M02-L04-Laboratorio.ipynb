## 5. `if`, `elif` e `else`

Python testa as condições de cima para baixo. Quando encontra a primeira condição verdadeira, executa aquele bloco.

````python
nota = 7

if nota >= 9:
    print("Excelente")
elif nota >= 6:
    print("Aprovado")
else:
    print("Precisa revisar")
````

## 6. Indentação: dentro e depois do bloco

No primeiro exemplo, duas linhas estão dentro do `if`. A última está depois do bloco.

````python
idade = 20

if idade >= 18:
    print("Condição verdadeira")
    print("Esta linha está dentro do if")

print("Esta linha está depois do if")
````

Agora observe como uma linha sem indentação não depende da condição.

````python
idade = 15

if idade >= 18:
    print("Esta linha só aparece se a condição for verdadeira")

print("Esta linha aparece depois da decisão")
````

### Exercício 5 — Reorganize a indentação

O trecho abaixo representa a intenção: mostrar duas mensagens **somente quando** a nota for maior ou igual a 6, e depois mostrar `"Fim"` independentemente da nota.

```text
nota = 7

if nota >= 6:
print("Aprovado")
print("Pode continuar")

    print("Fim")
```

Na célula seguinte, reescreva o código com a indentação correta. Depois, explique em uma frase por que a indentação altera o fluxo.

````python
nota = 7

if nota >= 6:
  print("Aprovado")
  print("Pode continuar")

print("Fim")
````

**Minha explicação sobre a indentação:** como as duas mensagens é para serem mostradas somente quando a nota for igual ou maior que 6, então fazemos a indentação desses prints. Mantemos o print("Fim") no começo da linha, pois ele vai ser executado independente do valor.

## 7. Repetição com `for`

Antes de executar, acompanhe mentalmente:

**pega um item → executa o bloco → pega o próximo item → executa → ...**

````python
nomes = ["Ana", "Bruno", "Carla"]

for nome in nomes:
    print("Nome atual:", nome)

print("Fim da lista")
````

## 8. `for` e `range()`

Compare os dois exemplos.

````python
nomes = ["Ana", "Bruno", "Carla"]

print("Percorrendo os itens da lista:")
for nome in nomes:
    print(nome)

print("Percorrendo números produzidos por range:")
for numero in range(3):
    print(numero)
````

### Exercício 7 — Quando usar cada forma?

Responda em texto:

1. Se você quer apenas imprimir cada nome de uma lista, por que `for nome in nomes` é mais direto do que gerar índices com `range()`?
2. Complete um exemplo que use `range()` para mostrar os números de 1 a 5.
3. Em uma frase, explique a diferença entre o papel de `for` e o de `range()`.

**Resposta 1:** pois com o range(), você teria que pegar a lista de nomes e o índice do range: **nomes[numero]**. Com o for nome in nomes é mais prático, com a váriavel nome: **print(nome)**

**Resposta 3:** A diferença é que o for é um laço de repetição, ou seja, vai repetir x vezes uma ação. O range, como o nome diz, é um intervalo de x até y valor, na forma de uma lista.

````python
for numero in range(1, 6):
  print(numero)
````

## 9. Contador

Um contador começa com um valor e é atualizado quando algo acontece.

````python
idades = [15, 22, 18, 14]
quantidade = 0

for idade in idades:
    if idade >= 18:
        quantidade = quantidade + 1

print("Quantidade de maiores de idade:", quantidade)
````

### Exercício 8 — Conte itens

Conte quantos valores da lista são maiores que 10.

````python
valores = [4, 12, 20, 7, 15]
quantidade = 0

for valor in valores:
  if valor > 10:
    quantidade = quantidade + 1

print("Quantidade de valores maiores que 10:", quantidade)
````

## 10. Acumulador

Agora observe uma variável que guarda uma soma ao longo do laço.

````python
valores = [10, 20, 5]
total = 0

for valor in valores:
    total = total + valor

print("Total:", total)
````

### Exercício 9 — Some valores de uma lista

````python
numeros = [3, 8, 2, 6]
total = 0

for numero in numeros:
    total = total + numero

print("Total:", total)
````

## 11. Lista de dicionários

Cada item percorrido pelo `for` será um dicionário.

````python
amostras = [
    {"id": "img_001", "classe": "A", "medida": 0.75},
    {"id": "img_002", "classe": "B", "medida": 0.42},
    {"id": "img_003", "classe": "A", "medida": 0.91}
]

for amostra in amostras:
    if amostra["medida"] > 0.70:
        print(amostra["id"], "acima do limite")
````
