## 4. A ordem de execução pode importar

Algumas células dependem de informações criadas anteriormente.

Vamos usar duas células para observar isso.

**Importante:** a linha com `mensagem = ...` usa uma variável. Variáveis serão estudadas formalmente na próxima lição. Aqui, esse exemplo serve apenas para mostrar a dependência entre células.

Execute primeiro a célula A e depois a célula B.

### Célula A — cria uma informação

````python
mensagem = "Esta informação foi criada na célula anterior."
````

### Célula B — usa a informação criada

````python
print(mensagem)
````

### Observação guiada

Se o ambiente estivesse novo e você executasse apenas a célula B, ela não encontraria a informação chamada `mensagem`.

Para iniciantes, uma boa prática é executar o notebook **de cima para baixo**.

**Resposta curta:** por que, neste exemplo, a célula A precisa ser executada antes da célula B?

**Sua resposta:**

A célula A precisa ser executada primeira, pois ela contém a atribuição de uma váriavel, que a célula B depende dessa váriavel para ser executada.
