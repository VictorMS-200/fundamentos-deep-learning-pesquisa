## 3. Exemplo guiado 2 — Balanceamento de classes

## 4. Exemplo guiado 3 — O Iris como lista de dicionários

Sem usar nenhuma biblioteca externa, podemos representar o dataset Iris como uma **lista de dicionários**, em que cada dicionário é uma amostra. Execute a célula abaixo e observe a estrutura.


````python
iris = [
    {"comprimento_sepala_cm": 5.1, "largura_sepala_cm": 3.5, "comprimento_petala_cm": 1.4, "largura_petala_cm": 0.2, "especie": "setosa"},
    {"comprimento_sepala_cm": 4.9, "largura_sepala_cm": 3.0, "comprimento_petala_cm": 1.4, "largura_petala_cm": 0.2, "especie": "setosa"},
    {"comprimento_sepala_cm": 4.7, "largura_sepala_cm": 3.2, "comprimento_petala_cm": 1.3, "largura_petala_cm": 0.2, "especie": "setosa"},
    {"comprimento_sepala_cm": 7.0, "largura_sepala_cm": 3.2, "comprimento_petala_cm": 4.7, "largura_petala_cm": 1.4, "especie": "versicolor"},
    {"comprimento_sepala_cm": 6.4, "largura_sepala_cm": 3.2, "comprimento_petala_cm": 4.5, "largura_petala_cm": 1.5, "especie": "versicolor"},
    {"comprimento_sepala_cm": 5.5, "largura_sepala_cm": 2.3, "comprimento_petala_cm": 4.0, "largura_petala_cm": 1.3, "especie": "versicolor"},
    {"comprimento_sepala_cm": 6.3, "largura_sepala_cm": 3.3, "comprimento_petala_cm": 6.0, "largura_petala_cm": 2.5, "especie": "virginica"},
    {"comprimento_sepala_cm": 5.8, "largura_sepala_cm": 2.7, "comprimento_petala_cm": 5.1, "largura_petala_cm": 1.9, "especie": "virginica"},
    {"comprimento_sepala_cm": 6.5, "largura_sepala_cm": 3.0, "comprimento_petala_cm": 5.8, "largura_petala_cm": 2.2, "especie": "virginica"},
    {"comprimento_sepala_cm": 6.0, "largura_sepala_cm": 2.7, "comprimento_petala_cm": 4.8, "largura_petala_cm": 1.6, "especie": "virginica"},
]

print(f"Número de amostras: {len(iris)}")
print("Primeira amostra:", iris[0])
````

Agora vamos separar conceitualmente **features (`X`)** e **rótulo (`y`)**, percorrendo a lista com um `for`.


````python
X = []
y = []

for amostra in iris:
    features = {
        "comprimento_sepala_cm": amostra["comprimento_sepala_cm"],
        "largura_sepala_cm": amostra["largura_sepala_cm"],
        "comprimento_petala_cm": amostra["comprimento_petala_cm"],
        "largura_petala_cm": amostra["largura_petala_cm"],
    }
    X.append(features)
    y.append(amostra["especie"])

print("X[0] (features):", X[0])
print("y[0] (rótulo):", y[0])
print("Total em X:", len(X), "| Total em y:", len(y))
````

Por fim, vamos contar quantas amostras existem de cada espécie usando um dicionário de contagem.

````python
contagem = {}

for especie in y:
    if especie in contagem:
        contagem[especie] += 1
    else:
        contagem[especie] = 1

print("Contagem por espécie:", contagem)
````

## 5. Exemplo guiado 4 — O classificador manual por regras `if/elif/else`

A função abaixo tenta prever a espécie de uma flor usando apenas `comprimento_petala_cm` e `largura_petala_cm`, com limiares fixos escolhidos por observação — **sem nenhum aprendizado a partir de dados**.


````python
def classificar(amostra):
    petala_c = amostra["comprimento_petala_cm"]
    petala_l = amostra["largura_petala_cm"]

    if petala_c < 2.5:
        return "setosa"
    elif petala_c < 5.0 and petala_l < 1.7:
        return "versicolor"
    else:
        return "virginica"


acertos = 0
erros = 0

for amostra in iris:
    predicao = classificar(amostra)
    real = amostra["especie"]
    status = "OK" if predicao == real else "ERRO"
    if predicao == real:
        acertos += 1
    else:
        erros += 1
    print(f"real={real:12s} predito={predicao:12s} [{status}]")

print()
print(f"Total de acertos: {acertos}")
print(f"Total de erros: {erros}")
````

  - Pois pode ter uma amostra com um parâmetro que intercepta a regra de outra amostra, fazendo com que o classificador desse erro no seu preditor.

## 7. Exemplo guiado 6 — Tipos de aprendizagem

* **Supervisionada**: prever a espécie de uma flor Iris sabendo a espécie correta de cada flor de treino.
* **Não supervisionada**: agrupar flores Iris por semelhança de medidas, sem informar a espécie de nenhuma.
* **Semi-supervisionada**: ter a espécie de poucas flores e usar as demais, sem rótulo, para ajudar a refinar o modelo.
* **Auto-supervisionada**: esconder uma medida de cada flor e treinar um modelo para prevê-la a partir das demais.
* **Por reforço**: um agente aprende jogando um jogo simples, recebendo pontos ao vencer e perdendo pontos ao errar.
